```python
import pm4py
import random
from pathlib import Path

input_path = Path("/Users/anyismacbook/Library/Mobile Documents/com~apple~CloudDocs/FSS26/Team Project/Conformance /Packing_Process.xes")
output_path = Path("/Users/anyismacbook/Library/Mobile Documents/com~apple~CloudDocs/FSS26/Team Project/Conformance /Packing_Process_test002.xes")
sample_ratio = 0.02

log = pm4py.read_xes(str(input_path))

num_traces = len(log)
sample_size = max(1, int(num_traces * sample_ratio))
sampled_indices = sorted(random.sample(range(num_traces), sample_size))
sampled_log = log.iloc[sampled_indices].reset_index(drop=True)

pm4py.write_xes(sampled_log, str(output_path))
print(f"Original: {num_traces} traces -> Sampled: {sample_size} traces ({sample_ratio*100}%)")
print(f"Output: {output_path}")
```
